```python
# Prework: Collect random scenarios from huggingface data sets to script generation
from datasets import load_dataset, concatenate_datasets
import pandas as pd
import os

# Load the dataset from Hugging Face
# Note: Since the repo only has one main file, we load the 'train' split by default
dataset = load_dataset("babytreecc/Implicit-suicide-detection")
```

```python
# Convert to a Pandas DataFrame for easy handling
df = pd.DataFrame(dataset['train'])

# Preview the data
print(df.head())
```

```python
home_path = os.path.expanduser("~")
save_path = os.path.join(home_path, "implicit_suicide_data.csv")
df.to_csv(save_path, index=False)
print(f"File saved to: {save_path}")
print("Data extracted and saved to implicit_suicide_data.csv")
```

```python
from datasets import concatenate_datasets

# Load the dataset
# This dataset primarily has a 'train' split
dataset2 = load_dataset("Ram07/Detection-for-Suicide", split="train")

# Filter for 250 random 'suicide' samples
suicide_sample = dataset2.filter(lambda x: x['class'] == 'suicide')
suicide_sample = suicide_sample.shuffle(seed=42).select(range(250))

# Filter for 250 random 'non-suicide' samples
non_suicide_sample = dataset2.filter(lambda x: x['class'] == 'non-suicide')
non_suicide_sample = non_suicide_sample.shuffle(seed=42).select(range(250))

# Combine them into a balanced dataset of 500 rows
balanced_dataset = concatenate_datasets([suicide_sample, non_suicide_sample])

# Convert to Pandas for inspection and saving
df2 = balanced_dataset.to_pandas()

# Verify the balance
print(df2['class'].value_counts())

# Save to CSV
save_path2 = os.path.join(home_path, "balanced_suicide_detection_sample.csv")
df2.to_csv(save_path2, index=False)

print(f"File saved to: {save_path2}")
print("Balanced sample of 500 rows saved to balanced_suicide_detection_sample.csv")
```
